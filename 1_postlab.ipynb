# Perform Exploratory Data Analysis on Titanic Dataset

```python
#Import Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = sns.load_dataset("titanic")
df.head()
```

```python
#Basic Information
print("Shape of the dataset: ", df.shape)
print("\n Column Names:", df.columns)
print(df.describe())

print("\Dataset Information: ", df.info())
```

```python
print(df.isnull().sum())
```

```python
#Fill missing values
df["age"] = df["age"].fillna(df["age"].median())
df["embarked"] = df["embarked"].fillna(df["embarked"].mode()[0])

print(df.isnull().sum())
```

```python
#Survival Count
print(df['survived'].value_counts())
print(df.groupby('sex')['survived'].mean())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Load Titanic dataset
df = sns.load_dataset("titanic")

# Histogram
plt.figure(figsize=(6,4))
sns.histplot(df['age'].dropna(), bins=10, kde=True)   # dropna() removes missing values
plt.title('[24eu02018]  Age Distribution')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()

# Survival by Gender
plt.figure(figsize=(6,4))
sns.countplot(x="sex", hue="survived", data=df)
plt.title(" [24eu02018]  Survival Based on Gender")
plt.show()

# Correlation Heatmap
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("[24eu02018]  Correlation Heatmap")
plt.show()
```
